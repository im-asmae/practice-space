```python
import pandas as pd 
import numpy as np 
import re
```

```python
doc1='Paris is the most beatiful city in europe'
doc2='i love visiting paris'
doc3='paris is one of the most popular tourist destinations in the world'
```

re.sub(motif, remplacement, texte) prend trois arguments :
1. Le motif à rechercher.
2. Ce par quoi le remplacer : ici, un espace " ".
3. Le texte à traiter.

```python
pdoc1=re.sub(r"[^a-zA-Z]", " ", doc1.lower()).split()
pdoc2=re.sub(r"[^a-zA-Z]", " ", doc2.lower()).split()
pdoc3=re.sub(r"[^a-zA-Z]", " ", doc3.lower()).split()
```

```python
print(pdoc1)
```

```python
wordset=np.union1d(pdoc1, pdoc2)
wordset=np.union1d(wordset, pdoc3)

print(wordset)
```

```python
# document-term matrix
# from the wordset list we create a dict to count the occurences of each word

def toBOW(wordset, pdoc):
    tf_dict = dict.fromkeys(wordset, 0)

    for word in pdoc:
        tf_dict[word] = pdoc.count(word)
        
    return tf_dict
```

```python
bow1=toBOW(wordset, pdoc1)
bow2=toBOW(wordset, pdoc2)
bow3=toBOW(wordset, pdoc3)

df_bow = pd.DataFrame([bow1, bow2, bow3])
df_bow
```

```python
df_bow.index = ["doc1", "doc2", "doc3"]
df_bow.index.name = "Documents"

df_bow
```

#### Implémentation de BoW avec scikit-learn

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
X = vectorizer.fit_transform([doc1, doc2, doc3])

print(vectorizer.get_feature_names_out())
```

```python
df_bow_sklearn= pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
df_bow_sklearn
```

```python
#si on ne voulait pas prendre en compte les stop words (mots vides) on peut passer ce parametre à CountVectorizer()

vectorizer2=CountVectorizer(stop_words='english')
X2=vectorizer2.fit_transform([doc1, doc2, doc3])

print(vectorizer2.get_feature_names_out())
```

```python
df_bow_sklearn_2=pd.DataFrame(X2.toarray(), columns=vectorizer2.get_feature_names_out())
df_bow_sklearn_2
```

Exercice d’application :

 Phrase N° 1: Il faut assurer un très bon stage de PFE

 Phrase N° 2 : Un très bon stage de PFE garantit un bon début de carrière

2. Convertir les deux phrases ci-dessus en vecteurs numériques de type BOW avec Python pure, puis

```python
#1. python pure
doc1='Il faut assurer un très bon stage de PFE'
doc2='Un très bon stage de PFE garantit un bon début de carrière'

pdoc1 = re.sub(r"[^a-zA-Z]", " ", doc1.lower()).split()
pdoc2 = re.sub(r"[^a-zA-Z]", " ", doc2.lower()).split()

vocab=np.union1d(pdoc1, pdoc2)

bow1=toBOW(vocab, pdoc1)
bow2=toBOW(vocab, pdoc2)

tf_bow=pd.DataFrame([bow1, bow2])

tf_bow
```

```python
# 2. avec sklearn

doc1='Il faut assurer un très bon stage de PFE'
doc2='Un très bon stage de PFE garantit un bon début de carrière'

vectorizer3=CountVectorizer()
X3=vectorizer3.fit_transform([doc1, doc2])

df_bow_sklearn_3=pd.DataFrame(X3.toarray(), columns=vectorizer3.get_feature_names_out())
df_bow_sklearn_3
```

**Conclusion :**

La comparaison des deux méthodes montre que le prétraitement du texte influence la construction du Bag of Words. L'implémentation en Python pur, utilisant l'expression régulière `r"[^a-zA-Z]"`, ne prend pas en charge les caractères accentués et peut donc découper incorrectement certains mots, comme « très ». En revanche, Scikit-Learn, avec `CountVectorizer`, reconnaît « très » comme un seul token. Cette comparaison met en évidence l'importance du choix de la méthode de tokenisation dans la représentation numérique des textes.
